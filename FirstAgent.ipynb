```python
from dotenv import load_dotenv
import os

load_dotenv(override=True)
```

```python
from crewai import LLM, Agent, Task, Crew

llm = LLM(
    model="openrouter/openai/gpt-4o-mini",       
    base_url="https://openrouter.ai/api/v1",      
    api_key=os.getenv("API_TOKEN"),        
)
```

```python
researcher = Agent(
    role="Senior Research Analyst",                   
    goal="Find and analyze the latest AI trends",    
    backstory="""You are a seasoned research analyst 
    with 10 years of experience in technology research. 
    You are known for your thorough analysis and 
    ability to spot emerging trends before they 
    become mainstream.""",                             
    llm=llm,                                         
    verbose=True    
)
```

```python
task = Task(
    title="Research",
    description="Preform a research on the history of gemini",
    expected_output="I need a detailed report on gemini report",
    agent=researcher
)
```

```python
crew = Crew(
    agents=[researcher],
    tasks=[task],
    verbose=True
)
```

```python
result = crew.kickoff()
print(result)
```
